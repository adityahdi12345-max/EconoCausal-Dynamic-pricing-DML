```python

import pandas as pd
import numpy as np

df = pd.read_excel("Online Retail.xlsx", engine="openpyxl")

print("Dataset loaded successfully!")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

display(df.head())
```

```python

# Check missing values in every column
missing = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing,
    "Missing Percentage (%)": (missing / len(df) * 100).round(2)
})

display(missing_report)
```

```python

print("Duplicate rows:", df.duplicated().sum())

print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())

print("Zero unit prices:", (df["UnitPrice"] == 0).sum())
print("Negative unit prices:", (df["UnitPrice"] < 0).sum())

print("\nQuantity summary:")
display(df["Quantity"].describe())

print("\nUnitPrice summary:")
display(df["UnitPrice"].describe())
```

```python

# Inspect transactions with negative quantities
negative_qty = df[df["Quantity"] < 0]

print("Negative quantity records:", len(negative_qty))

display(
    negative_qty[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice"]
    ].head(10)
)

# Check invoice numbers that begin with C
cancelled = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
]

print("Invoices with C prefix:", len(cancelled))
display(cancelled.head(10))
```

```python

zero_price = df[df["UnitPrice"] == 0]

print("Zero-price records:", len(zero_price))

display(
    zero_price[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "Country"]
    ].head(10)
)
```

```python

# Check missing descriptions among zero-price records
zero_price = df[df["UnitPrice"] == 0].copy()

print("Total zero-price records:", len(zero_price))
print(
    "Missing descriptions among zero-price records:",
    zero_price["Description"].isna().sum()
)

# Check whether these records have customer IDs
print(
    "Missing CustomerID among zero-price records:",
    zero_price["CustomerID"].isna().sum()
)

# View records that have a description
display(
    zero_price[zero_price["Description"].notna()][
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "CustomerID", "Country"]
    ].head(10)
)
```

```python

zero_price = df[df["UnitPrice"] == 0]

print("Positive quantities:", (zero_price["Quantity"] > 0).sum())
print("Negative quantities:", (zero_price["Quantity"] < 0).sum())
print("Missing descriptions:", zero_price["Description"].isna().sum())
print("Missing CustomerID:", zero_price["CustomerID"].isna().sum())

display(
    zero_price.groupby("Quantity").size()
    .sort_values(ascending=False)
    .head(10)
)
```

```python

# Display sample duplicate rows
duplicates = df[df.duplicated(keep=False)]

print("Rows involved in duplicates:", len(duplicates))
display(duplicates.head(10))
```

```python

# Preserve the original dataset
df_clean = df.copy()

# Remove only exact duplicate rows
before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

after = len(df_clean)

print("Original rows:", before)
print("Rows after removing exact duplicates:", after)
print("Exact duplicate rows removed:", before - after)
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

```python

print("Missing values after duplicate removal:")
display(df_clean.isnull().sum().to_frame("Missing Count"))

print("Cleaned dataset shape:", df_clean.shape)
```

```python

from pathlib import Path
import pandas as pd

Path("reports").mkdir(exist_ok=True)

cleaning_report = pd.DataFrame({
    "Metric": [
        "Original rows",
        "Exact duplicate rows removed",
        "Rows after duplicate removal",
        "Remaining exact duplicates",
        "Missing Description values",
        "Missing CustomerID values",
        "Zero-price records",
        "Negative-quantity records"
    ],
    "Value": [
        len(df),
        len(df) - len(df_clean),
        len(df_clean),
        df_clean.duplicated().sum(),
        df_clean["Description"].isna().sum(),
        df_clean["CustomerID"].isna().sum(),
        (df_clean["UnitPrice"] == 0).sum(),
        (df_clean["Quantity"] < 0).sum()
    ]
})

display(cleaning_report)

cleaning_report.to_csv(
    "reports/day6_cleaning_summary.csv",
    index=False
)

print("Cleaning summary saved successfully!")
```
