```python
import pandas as pd
import numpy as np

from google.colab import files
from sklearn.model_selection import KFold
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge, LinearRegression

print("Libraries imported successfully!")
```

```python
df = pd.read_excel("Online Retail.xlsx")
df = df.drop_duplicates().copy()

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)

data = df[
    (df["UnitPrice"] > 0) &
    (df["Quantity"] > 0) &
    (df["InvoiceDate"].notna()) &
    (df["StockCode"].notna())
].copy()

data["Month"] = data["InvoiceDate"].dt.month
data["Hour"] = data["InvoiceDate"].dt.hour
data["DayOfWeek"] = data["InvoiceDate"].dt.dayofweek

print("Prepared rows:", len(data))
print(data.head())
```

```python
sample = data[
    ["UnitPrice", "Quantity", "StockCode",
     "Month", "Hour", "DayOfWeek"]
].dropna().copy()

if len(sample) > 30000:
    sample = sample.sample(n=30000, random_state=42)

T = sample["UnitPrice"].to_numpy()
Y = sample["Quantity"].to_numpy()

X = sample[["StockCode", "Month", "Hour", "DayOfWeek"]].copy()

print("Sample size:", len(sample))
print("Unique products:", X["StockCode"].nunique())
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "product",
            OneHotEncoder(handle_unknown="ignore"),
            ["StockCode"]
        ),
        (
            "time",
            "passthrough",
            ["Month", "Hour", "DayOfWeek"]
        )
    ]
)

kf = KFold(n_splits=2, shuffle=True, random_state=42)

T_residual = np.zeros(len(sample))
Y_residual = np.zeros(len(sample))

for train_idx, test_idx in kf.split(X):
    model_t = make_pipeline(
        preprocessor,
        Ridge(alpha=10.0)
    )

    model_y = make_pipeline(
        preprocessor,
        Ridge(alpha=10.0)
    )

    model_t.fit(X.iloc[train_idx], T[train_idx])
    model_y.fit(X.iloc[train_idx], Y[train_idx])

    T_residual[test_idx] = (
        T[test_idx] - model_t.predict(X.iloc[test_idx])
    )

    Y_residual[test_idx] = (
        Y[test_idx] - model_y.predict(X.iloc[test_idx])
    )

denominator = np.mean(T_residual ** 2)

if denominator <= 1e-10:
    raise ValueError(
        "Insufficient residual price variation for this prototype."
    )

theta = np.mean(T_residual * Y_residual) / denominator

print("Preliminary DML coefficient:", round(theta, 6))
print("Residual price variation:", round(denominator, 6))
```
