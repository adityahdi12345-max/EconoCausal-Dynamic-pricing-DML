```python
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_excel("Online Retail.xlsx")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
# Work on a copy; keep the original data unchanged
data = df.drop_duplicates().copy()

# Convert date column to datetime
data["InvoiceDate"] = pd.to_datetime(
    data["InvoiceDate"], errors="coerce"
)

# Create time-based features
data["InvoiceYear"] = data["InvoiceDate"].dt.year
data["InvoiceMonth"] = data["InvoiceDate"].dt.month
data["InvoiceDay"] = data["InvoiceDate"].dt.day
data["InvoiceHour"] = data["InvoiceDate"].dt.hour
data["DayOfWeek"] = data["InvoiceDate"].dt.dayofweek

# Create a recorded line-value feature
data["LineValue"] = data["Quantity"] * data["UnitPrice"]

print("New dataset shape:", data.shape)
print("\nNew features:")
print([
    "InvoiceYear", "InvoiceMonth", "InvoiceDay",
    "InvoiceHour", "DayOfWeek", "LineValue"
])

data.head()
```

```python
print("Missing dates:", data["InvoiceDate"].isna().sum())
print("Non-positive quantity:", (data["Quantity"] <= 0).sum())
print("Non-positive price:", (data["UnitPrice"] <= 0).sum())

print("\nFeature summary:")
data[
    ["InvoiceMonth", "InvoiceHour", "DayOfWeek", "LineValue"]
].describe()
```
