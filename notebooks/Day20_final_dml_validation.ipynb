```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Day 15 preliminary DML results
results = pd.DataFrame({
    "Model": [
        "Time controls only",
        "Product and time controls"
    ],
    "Coefficient": [
        -0.005836,
        -0.002905
    ]
})

results
```

```python
# Calculate coefficient difference

coefficient_difference = (
    results.loc[1, "Coefficient"] -
    results.loc[0, "Coefficient"]
)

print("Time-only model:", results.loc[0, "Coefficient"])
print("Product + time model:", results.loc[1, "Coefficient"])
print("Coefficient difference:", coefficient_difference)
```

```python
# Plot coefficient comparison

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["Coefficient"]
)

plt.title("DML Coefficient Comparison")
plt.xlabel("Model Specification")
plt.ylabel("Preliminary Coefficient")

plt.axhline(0, color="black", linewidth=0.8)
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```
